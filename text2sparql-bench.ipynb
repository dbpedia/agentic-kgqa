```python
from __future__ import annotations

from pathlib import Path
from rdflib import Graph
from rdflib.exceptions import ParserError
from glob import glob

def stream_check_ntriples(path: str | Path, *, chunk_lines: int = 50_000, encoding: str = "utf-8"):
    """
    Stream-validate N-Triples (.nt) or N-Quads (.nq) by parsing in line chunks.
    Returns a list of (chunk_index, start_line, end_line, error_message).
    """
    path = Path(path)
    fmt = "nquads" if path.suffix.lower() == ".nq" else "nt"

    errors = []
    buf = []
    start_line = 1
    chunk_idx = 0

    # rdflib parses from bytes/str; we feed it chunk text
    with path.open("r", encoding=encoding, errors="replace", newline="") as f:
        for lineno, line in enumerate(f, start=1):
            # skip empty lines / comments (common in .nt)
            stripped = line.lstrip()
            if not stripped or stripped.startswith("#"):
                continue

            buf.append(line)
            if len(buf) >= chunk_lines:
                chunk_idx += 1
                g = Graph()
                data = "".join(buf)
                try:
                    g.parse(data=data, format=fmt)
                except (ParserError, Exception) as e:
                    errors.append((chunk_idx, start_line, lineno, str(e)))
                buf.clear()
                start_line = lineno + 1

        # final chunk
        if buf:
            chunk_idx += 1
            g = Graph()
            data = "".join(buf)
            try:
                g.parse(data=data, format=fmt)
            except (ParserError, Exception) as e:
                errors.append((chunk_idx, start_line, lineno, str(e)))

    return errors

for rdf_file in glob("/Users/tom/graphdb-import/*"):
    print(f"Checking: {rdf_file}")
    errs = stream_check_ntriples(rdf_file, chunk_lines=100_000)
    if not errs:
        print("✅ No parse errors found.")
    else:
        print(f"❌ Found {len(errs)} chunk(s) with errors:")
        for chunk_idx, a, b, msg in errs:
            print(f"- chunk {chunk_idx} (lines ~{a}-{b}): {msg}")
```
